## 13.1 Write Current Date to today.txt

```python
# 13.1 Write the current date as a string to the text file today.txt.
from datetime import datetime

# Get today's date formatted as YYYY-MM-DD
current_date = datetime.now().strftime("%Y-%m-%d")
print(f"Current date: {current_date}")

# Write it to today.txt
with open("today.txt", "w") as file:
    file.write(current_date)

print("✅ Wrote to today.txt")
```

```python
# Verify the file exists in the current directory
import os
print(os.listdir("."))   # Should include 'today.txt'
```

## 13.2 Read today.txt into today_string

```python
# 13.2 Read the text file today.txt into the string today_string.
with open("today.txt", "r") as file:
    today_string = file.read()

print(f"today_string = '{today_string}'")
print(f"Type: {type(today_string)}")
```

## 13.3 Parse the Date from today_string

```python
# 13.3 Parse the date from today_string.
from datetime import datetime

# Convert the string back into a datetime object
parsed_date = datetime.strptime(today_string, "%Y-%m-%d")

print(f"Parsed: {parsed_date}")
print(f"Type:   {type(parsed_date)}")
print(f"Year:   {parsed_date.year}")
print(f"Month:  {parsed_date.month}")
print(f"Day:    {parsed_date.day}")
```
