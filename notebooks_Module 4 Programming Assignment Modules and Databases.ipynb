### 16.4
Used `sqlite3` to create `books.db` and a `books` table with columns
`title` (TEXT), `author` (TEXT), and `year` (INTEGER).

```python
import sqlite3

conn = sqlite3.connect('books.db')
curs = conn.cursor()

curs.execute('''CREATE TABLE IF NOT EXISTS books
                (title TEXT,
                 author TEXT,
                 year INTEGER)''')
conn.commit()
conn.close()
print("books.db created with 'books' table.")
```

### 16.5
Read `books2.csv` with `csv.DictReader` and inserted all rows into the `books`
table using `executemany()`, converting `year` to `int`.

```python
import csv, sqlite3

conn = sqlite3.connect('books.db')
curs = conn.cursor()

# Clear table so re-running this cell doesn't duplicate rows
curs.execute('DELETE FROM books')

with open('books2.csv', newline='') as fin:
    reader = csv.DictReader(fin)
    rows = [(r['title'], r['author'], int(r['year'])) for r in reader]

curs.executemany('INSERT INTO books (title, author, year) VALUES (?, ?, ?)', rows)
conn.commit()

for row in curs.execute('SELECT * FROM books'):
    print(row)

conn.close()
```

### 16.8
Connected to `books.db` via SQLAlchemy (`create_engine('sqlite:///books.db')`)
and selected the `title` column ordered alphabetically.

```python
%pip install sqlalchemy
```

```python
from sqlalchemy import create_engine, text

engine = create_engine('sqlite:///books.db', echo=False)

with engine.connect() as conn:
    result = conn.execute(text('SELECT title FROM books ORDER BY title ASC'))
    for (title,) in result:
        print(title)
```
