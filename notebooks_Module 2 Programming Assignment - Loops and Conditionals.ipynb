## Chapter 4: Conditional Statements

This section demonstrates if/elif/else statements with number comparisons and logical operators.

```python
# 4.2 - Fruit Matcher with if/elif/else
print("4.2 - Fruit Matcher")
print("-" * 30)

# Test Case 1: Small and Green
print("Test Case 1: small=True, green=True")
small = True
green = True

if small and green:
    print("Result: pea")
elif small and not green:
    print("Result: cherry")
elif not small and green:
    print("Result: watermelon")
else:
    print("Result: pumpkin")
print()

# Test Case 2: Small but not Green
print("Test Case 2: small=True, green=False")
small = True
green = False

if small and green:
    print("Result: pea")
elif small and not green:
    print("Result: cherry")
elif not small and green:
    print("Result: watermelon")
else:
    print("Result: pumpkin")
print()

# Test Case 3: Not Small but Green
print("Test Case 3: small=False, green=True")
small = False
green = True

if small and green:
    print("Result: pea")
elif small and not green:
    print("Result: cherry")
elif not small and green:
    print("Result: watermelon")
else:
    print("Result: pumpkin")
print()

# Test Case 4: Neither Small nor Green
print("Test Case 4: small=False, green=False")
small = False
green = False

if small and green:
    print("Result: pea")
elif small and not green:
    print("Result: cherry")
elif not small and green:
    print("Result: watermelon")
else:
    print("Result: pumpkin")
print()

print("=" * 40)
print("MATCHING LOGIC:")
print("- small=True, green=True  → pea")
print("- small=True, green=False → cherry")
print("- small=False, green=True → watermelon")
print("- small=False, green=False → pumpkin")
print("=" * 40)
```

## Chapter 6: Loops

This section demonstrates for loops, while loops, and loop control with break statements.

```python
# 6.1 - For Loop with List
print("6.1 - For Loop with List")
print("-" * 30)

print("Printing values from list [3, 2, 1, 0]:")
for value in [3, 2, 1, 0]:
    print(value)

print()
```

```python
# 6.2 - While Loop with guess_me
print("6.2 - While Loop with guess_me")
print("-" * 30)

guess_me = 7
number = 1

print(f"Looking for: {guess_me}")
print("Starting from: 1")
print()

while True:
    if number < guess_me:
        print(f"{number}: too low")
    elif number == guess_me:
        print(f"{number}: found it!")
        break
    else:  # number > guess_me
        print(f"{number}: oops")
        break
    number += 1

print()
```

```python
# 6.3 - For Loop with range()
print("6.3 - For Loop with range(10)")
print("-" * 30)

guess_me = 5
print(f"Looking for: {guess_me}")
print()

for number in range(10):
    if number < guess_me:
        print(f"{number}: too low")
    elif number == guess_me:
        print(f"{number}: found it!")
        break
    else:  # number > guess_me
        print(f"{number}: oops")
        break

print()
```
