# Chapter 3 Exercises

```python
# 3.1 - Seconds in an hour
60 * 60
```

```python
# 3.2 - Assign to variable
seconds_per_hour = 60 * 60
seconds_per_hour
```

```python
# 3.3 - Seconds in a day
seconds_per_hour * 24
```

```python
# 3.4 - Save to seconds_per_day
seconds_per_day = seconds_per_hour * 24
seconds_per_day
```

```python
# 3.5 - Floating-point division
seconds_per_day / seconds_per_hour
```

```python
# 3.6 - Integer division
seconds_per_day // seconds_per_hour
```

# Answers
3.1: There are 3600 seconds in an hour.
3.2: seconds_per_hour = 3600
3.3: There are 86400 seconds in a day.
3.4: seconds_per_day = 86400
3.5: 24.0
3.6: 24. This agrees with the floating-point value (just without .0)
