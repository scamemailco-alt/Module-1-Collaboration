# Vehicle / Automobile App

A small OOP demo with a `Vehicle` superclass and an `Automobile` subclass that collects car details from the user.

## 1. Define the Vehicle superclass

`Vehicle` holds the general vehicle type (car, truck, plane, boat, broomstick, etc.). The subclass will call this constructor with the appropriate type.

```python
class Vehicle:
    def __init__(self, vehicle_type):
        self.vehicle_type = vehicle_type
```

## 2. Define the Automobile subclass

`Automobile` inherits from `Vehicle` and adds `year`, `make`, `model`, `doors`, and `roof`. It calls `super().__init__("car")` so the vehicle type is always stored as "car" for this class.

```python
class Automobile(Vehicle):
    def __init__(self, year, make, model, doors, roof):
        super().__init__("car")
        self.year = year
        self.make = make
        self.model = model
        self.doors = doors
        self.roof = roof
```

## 3. Define a display method

Add a method that prints the car details in the requested easy-to-read format.

```python
class Automobile(Vehicle):
    def __init__(self, year, make, model, doors, roof):
        super().__init__("car")
        self.year = year
        self.make = make
        self.model = model
        self.doors = doors
        self.roof = roof

    def display(self):
        print(f"Vehicle type: {self.vehicle_type}")
        print(f"Year: {self.year}")
        print(f"Make: {self.make}")
        print(f"Model: {self.model}")
        print(f"Number of doors: {self.doors}")
        print(f"Type of roof: {self.roof}")
```

## 4. Collect user input

Prompt the user for each attribute. I'll accept free-text for `doors` and `roof` so the user can type "2", "4", "solid", or "sun roof" as the assignment describes. A simple validation loop nudges them if they enter something unexpected.

```python
year = input("Enter the year: ").strip()
make = input("Enter the make: ").strip()
model = input("Enter the model: ").strip()

# Validate doors
while True:
    doors = input("Enter the number of doors (2 or 4): ").strip()
    if doors in ("2", "4"):
        break
    print("Please enter 2 or 4.")

# Validate roof
while True:
    roof = input("Enter the type of roof (solid or sun roof): ").strip().lower()
    if roof in ("solid", "sun roof"):
        break
    print("Please enter 'solid' or 'sun roof'.")
```

## 5. Create the Automobile and display it

Build the object from the collected input and call `display()`.

```python
my_car = Automobile(year, make, model, doors, roof)
my_car.display()
```
