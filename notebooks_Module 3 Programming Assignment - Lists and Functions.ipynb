## 7.5 Capitalize the person

```python
things[1] = things[1].capitalize()
print(things)
```

Did it change the element in the list? Yes. Strings are immutable, so capitalize() returns a new string, but I reassigned it back to things[1]. Since lists are mutable, the list now holds the capitalized version.

## 9.2 Define get_odds()

```python
def get_odds():
    for num in range(10):
        if num % 2 == 1:
            yield num
```

```python
count = 0
for odd in get_odds():
    count += 1
    if count == 3:
        print(odd)
        break
```

get_odds() is a generator because it uses yield. It produces 1, 3, 5, 7, 9. The for loop counts each value and prints the third one (5) before breaking.
